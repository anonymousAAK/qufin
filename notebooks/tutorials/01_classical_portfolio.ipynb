# Tutorial 1: Classical Portfolio Optimization

This notebook demonstrates classical mean-variance portfolio optimization using `qufin`.
We cover the Markowitz framework, efficient frontier construction, and Sharpe ratio maximization.

**Prerequisites**: `pip install qufin`

```python
import numpy as np
np.random.seed(42)
```

## 1. Define the Investment Universe

We create a 5-asset universe with expected returns and a covariance matrix.

```python
# Expected annual returns for 5 assets
mu = np.array([0.12, 0.10, 0.07, 0.03, 0.15])

# Covariance matrix (annualized)
cov = np.array([
    [0.040, 0.006, 0.002, 0.000, 0.010],
    [0.006, 0.030, 0.004, 0.001, 0.008],
    [0.002, 0.004, 0.020, 0.002, 0.003],
    [0.000, 0.001, 0.002, 0.010, 0.001],
    [0.010, 0.008, 0.003, 0.001, 0.050],
])

asset_names = ["Tech", "Healthcare", "Utilities", "Bonds", "Energy"]
print(f"Assets: {asset_names}")
print(f"Expected returns: {mu}")
```

## 2. Minimum Variance Portfolio

```python
from qufin.portfolio.classical.mean_variance import mean_variance, Objective

# Minimum variance portfolio
result_mv = mean_variance(mu, cov, objective=Objective.MIN_VARIANCE)

print("Minimum Variance Portfolio")
print(f"  Weights:  {np.round(result_mv.weights, 4)}")
print(f"  Return:   {result_mv.expected_return:.4f}")
print(f"  Vol:      {result_mv.volatility:.4f}")
print(f"  Sharpe:   {result_mv.sharpe_ratio:.4f}")
```

## 3. Maximum Sharpe Ratio Portfolio

```python
# Maximum Sharpe ratio portfolio
result_sharpe = mean_variance(
    mu, cov,
    objective=Objective.MAX_SHARPE,
    risk_free_rate=0.02,
)

print("Maximum Sharpe Portfolio")
print(f"  Weights:  {np.round(result_sharpe.weights, 4)}")
print(f"  Return:   {result_sharpe.expected_return:.4f}")
print(f"  Vol:      {result_sharpe.volatility:.4f}")
print(f"  Sharpe:   {result_sharpe.sharpe_ratio:.4f}")
```

## 4. Efficient Frontier

We trace the efficient frontier by solving for minimum variance at a range of target returns.

```python
target_returns = np.linspace(0.04, 0.14, 20)
frontier_vols = []
frontier_rets = []

for target in target_returns:
    try:
        res = mean_variance(
            mu, cov,
            objective=Objective.MIN_VARIANCE,
            min_return=target,
        )
        frontier_vols.append(res.volatility)
        frontier_rets.append(res.expected_return)
    except Exception:
        continue

print(f"Computed {len(frontier_vols)} points on the efficient frontier")
for r, v in zip(frontier_rets[:5], frontier_vols[:5]):
    print(f"  Return={r:.4f}  Vol={v:.4f}")
```

## 5. Risk Parity

Risk parity equalizes each asset's contribution to total portfolio risk.

```python
from qufin.portfolio.classical.risk_parity import risk_parity

rp_result = risk_parity(cov)

print("Risk Parity Portfolio")
print(f"  Weights: {np.round(rp_result.weights, 4)}")
print(f"  Vol:     {rp_result.volatility:.4f}")
```

## 6. Hierarchical Risk Parity (HRP)

```python
from qufin.portfolio.classical.hrp import hrp_optimize

hrp_result = hrp_optimize(cov)

print("HRP Portfolio")
print(f"  Weights: {np.round(hrp_result.weights, 4)}")
```
